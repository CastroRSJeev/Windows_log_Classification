```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import IsolationForest
```

```python
df = pd.read_csv('./data/modified.csv')
```

```python
df.info(0)
```

```python
df = df.dropna(subset=['process_name'])
df['user_context'] = df['user_context'].fillna('SYSTEM')
df['user_id'] = df['user_id'].fillna('unknown')
df['parent_process_name'] = df['parent_process_name'].fillna('unknown')
df['parent_process_path'] = df['parent_process_path'].fillna('unknown')
df['logon_type'] = df['logon_type'].fillna(0)
df['event_count_1h'] = df['event_count_1h'].fillna(0)
df['source_port'] = df['source_port'].fillna(0)
df['protocol'] = df['protocol'].fillna('none')
df['source_address'] = df['source_address'].fillna('none')
df['dest_address'] = df['dest_address'].fillna('none')
df['process_path'] = df['process_path'].fillna('unknown')
df['command_line'] = df['command_line'].fillna('none')
df['file_path'] = df['file_path'].fillna('none')
df['file_name'] = df['file_name'].fillna('none')
df['file_extension'] = df['file_extension'].fillna('none')
```

```python
df.head()
```

```python
df.info()
```

```python
df.to_csv('./data/new.csv')
```

```python
df.drop(columns=['day_of_week'],inplace=True)
```

```python
df.drop(columns=['parent_process_name','parent_process_path','command_line','source_address','dest_address','user_id','user_context','logon_type','source_port','protocol'],inplace=True)
```

```python
df.drop(columns=['Unnamed: 0'],inplace=True)
```

```python
df.info()
```

```python
le = LabelEncoder()
for col in ['source_ip_type', 'dest_ip_type', 'registry_path_type', 'hostname_role']:
    df[col] = le.fit_transform(df[col])
```

```python
df = pd.get_dummies(df, columns=['event_type'])
freq = df['process_name'].value_counts(normalize=True)
df['process_name_freq'] = df['process_name'].map(freq)
```

```python
def bin_port(port):
    if port <= 1023: return 'well_known'
    elif port <= 49151: return 'registered'
    else: return 'dynamic'

df['dest_port_category'] = df['dest_port'].apply(bin_port)
df = pd.get_dummies(df, columns=['dest_port_category'])
```

```python
risky_dirs = ['temp', 'downloads', 'appdata', 'public', 'desktop']
df['process_in_risky_dir'] = df['process_path'].str.lower().str.contains('|'.join(risky_dirs), na=False).astype(int)
```

```python
df.head()
```

```python
df.info()
```

```python
risky_dirs = ['temp', 'downloads', 'appdata', 'public', 'desktop']
df['file_in_risky_dir'] = df['file_path'].str.lower().str.contains('|'.join(risky_dirs), na=False).astype(int)

df['is_executable'] = df['file_name'].str.lower().str.endswith(('.exe', '.bat', '.ps1', '.vbs', '.cmd'), na=False).astype(int)

df.drop(columns=['process_name', 'hostname', 'process_path', 'file_path', 'file_name', 'file_extension'], inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
df.drop(columns=['hour_of_day', 'unique_destinations_1h', 'dest_port'], inplace=True)
```

```python
for i in df.columns:
    print(f"{i}: {len(df[i].unique())}")
```

```python
df['event_count_1h'] = pd.cut(df['event_count_1h'],
                               bins=[0, 10, 50, 100, float('inf')],
                               labels=[0, 1, 2, 3],
                               include_lowest=True).astype(int)
```

```python
df.info()
```

```python
df.head()
```

```python
df = df.sample(frac=1).reset_index(drop=True)
```

```python
df.head()
```

```python
df['dest_port_category_registered'] = df['dest_port_category_registered'].astype('int')
```

```python
df['dest_port_category_dynamic'] = df['dest_port_category_dynamic'].astype('int')
df['dest_port_category_well_known'] = df['dest_port_category_well_known'].astype('int')
```

```python
df['event_type_DeleteValue'] = df['event_type_DeleteValue'].astype('int')
df['event_type_INFO'] = df['event_type_INFO'].astype('int')
df['event_type_SetValue'] = df['event_type_SetValue'].astype('int')
```

```python
df.info()
```

```python
int((80/100)*len(df))
```

```python

X = df.copy()
model = IsolationForest(n_estimators=100, contamination=0.05, random_state=42)
df['anomaly_score'] = model.fit_predict(X)
df['isolation_score'] = model.decision_function(X)
```

```python
df.to_csv('./data/isolated.csv')
```
